```python
import pandas as pd
import numpy as np

application_bureau = pd.read_csv(
    "../data/processed/application_bureau.csv"
)

prev = pd.read_csv(
    "../data/raw/previous_application.csv"
)
```

```python
prev_features = prev.groupby("SK_ID_CURR").agg(

    prev_app_count=("SK_ID_PREV","count"),

    approved_count=(
        "NAME_CONTRACT_STATUS",
        lambda x:(x=="Approved").sum()
    ),

    refused_count=(
        "NAME_CONTRACT_STATUS",
        lambda x:(x=="Refused").sum()
    ),

    avg_prev_credit=(
        "AMT_CREDIT",
        "mean"
    ),

    max_prev_credit=(
        "AMT_CREDIT",
        "max"
    ),

    avg_prev_annuity=(
        "AMT_ANNUITY",
        "mean"
    )

).reset_index()
```

```python
prev_features["approval_ratio"] = (
    prev_features["approved_count"]
    /
    prev_features["prev_app_count"]
)

prev_features["refusal_ratio"] = (
    prev_features["refused_count"]
    /
    prev_features["prev_app_count"]
)
```

```python
prev_features = prev_features.replace(
    [np.inf,-np.inf],
    np.nan
)
```

```python
application_bureau_prev = (
    application_bureau.merge(
        prev_features,
        on="SK_ID_CURR",
        how="left"
    )
)
```

```python
application_bureau_prev.to_csv(
    "../data/processed/application_bureau_prev.csv",
    index=False
)
```
