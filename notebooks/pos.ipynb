```python
import pandas as pd
import numpy as np

application = pd.read_csv(
    "../data/processed/application_bureau_prev_inst.csv"
)

pos = pd.read_csv(
    "../data/raw/POS_CASH_balance.csv"
)
```

```python
pos["DPD_FLAG"] = (
    pos["SK_DPD"] > 0
).astype(int)

pos["DPD_30"] = (
    pos["SK_DPD"] > 30
).astype(int)

pos["DPD_90"] = (
    pos["SK_DPD"] > 90
).astype(int)
```

```python
pos_features = pos.groupby(
    "SK_ID_CURR"
).agg(

    pos_num_loans=(
        "SK_ID_PREV",
        "nunique"
    ),

    pos_oldest_month=(
        "MONTHS_BALANCE",
        "min"
    ),

    pos_latest_month=(
        "MONTHS_BALANCE",
        "max"
    ),

    pos_avg_installments=(
        "CNT_INSTALMENT",
        "mean"
    ),

    pos_avg_future_installments=(
        "CNT_INSTALMENT_FUTURE",
        "mean"
    ),

    pos_avg_dpd=(
        "SK_DPD",
        "mean"
    ),

    pos_max_dpd=(
        "SK_DPD",
        "max"
    ),

    pos_avg_dpd_def=(
        "SK_DPD_DEF",
        "mean"
    ),

    pos_max_dpd_def=(
        "SK_DPD_DEF",
        "max"
    ),

    pos_late_ratio=(
        "DPD_FLAG",
        "mean"
    ),

    pos_dpd30_ratio=(
        "DPD_30",
        "mean"
    ),

    pos_dpd90_ratio=(
        "DPD_90",
        "mean"
    )

).reset_index()
```

```python
pos_features = pos_features.replace(
    [np.inf, -np.inf],
    np.nan
)
```

```python
application_pos = application.merge(
    pos_features,
    on="SK_ID_CURR",
    how="left"
)
```

```python
application_pos.to_csv(
    "../data/processed/application_bureau_prev_inst_pos.csv",
    index=False
)
```
